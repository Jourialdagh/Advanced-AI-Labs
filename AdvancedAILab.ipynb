```python
from sklearn.preprocessing import LabelEncoder

# Encode 'Precip Type' into numeric labels
label_encoder = LabelEncoder()
df['Precip Type'] = label_encoder.fit_transform(df['Precip Type'])

print("\nUnique Precip Types after encoding:")
print(df['Precip Type'].unique())
print("\nData types after encoding:")
print(df.dtypes)
```

```python
# Drop irrelevant and non-numeric columns
df_processed = df.drop(['Formatted Date', 'Daily Summary'], axis=1)
```

```python
# Drop irrelevant and non-numeric columns
df_processed = df.drop(['Formatted Date', 'Daily Summary', 'Summary'], axis=1)

# Define features (X) and target (y)
# Assuming 'Apparent Temperature (C)' is the target variable based on common weather analysis tasks
# If your target is different, please specify.
X = df_processed.drop('Apparent Temperature (C)', axis=1)
y = df_processed['Apparent Temperature (C)']

# Apply feature scaling to the features (X)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("\nShape of X after scaling:", X_scaled.shape)
```

**Task 4: Ensemble Learning**

Split data into 80% training and 20% testing sets using train_test_split.

Define X (features) and y (target).

You will train and compare three classifiers:

RandomForestClassifier

GradientBoostingClassifier

XGBClassifier

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
import xgboost as xgb
from sklearn.metrics import accuracy_score, classification_report

# Split data into training and testing sets (80/20 split)
# Assuming the target variable is 'Precip Type' based on the previous steps which encoded it
X_train, X_test, y_train, y_test = train_test_split(X_scaled, df['Precip Type'], test_size=0.2, random_state=42)

print("Shape of X_train:", X_train.shape)
print("Shape of X_test:", X_test.shape)
print("Shape of y_train:", y_train.shape)
print("Shape of y_test:", y_test.shape)
```

```python
# Initialize the classifiers
rf_classifier = RandomForestClassifier(n_estimators=100, random_state=42)
gb_classifier = GradientBoostingClassifier(n_estimators=100, learning_rate=0.1, random_state=42)
xgb_classifier = xgb.XGBClassifier(n_estimators=100, learning_rate=0.1, eval_metric='mlogloss', random_state=42)

# Train the classifiers
print("Training RandomForestClassifier...")
rf_classifier.fit(X_train, y_train)

print("Training GradientBoostingClassifier...")
gb_classifier.fit(X_train, y_train)

print("Training XGBClassifier...")
xgb_classifier.fit(X_train, y_train)

print("Training complete.")
```

```python
# Make predictions on the test set
y_pred_rf = rf_classifier.predict(X_test)
y_pred_gb = gb_classifier.predict(X_test)
y_pred_xgb = xgb_classifier.predict(X_test)

# Evaluate the classifiers
print("RandomForestClassifier Performance:")
print("Accuracy:", accuracy_score(y_test, y_pred_rf))
print(classification_report(y_test, y_pred_rf))

print("\nGradientBoostingClassifier Performance:")
print("Accuracy:", accuracy_score(y_test, y_pred_gb))
print(classification_report(y_test, y_pred_gb))

print("\nXGBClassifier Performance:")
print("Accuracy:", accuracy_score(y_test, y_pred_xgb))
print(classification_report(y_test, y_pred_xgb))
```

**Task 5: Reflection**

Based on the results from Task 4, all three ensemble methods (Random Forest, Gradient Boosting, and XGBoost) achieved very high accuracy on this dataset for predicting 'Precip Type'. In this specific case, where all models performed nearly perfectly, it's difficult to definitively recommend one over the others based solely on accuracy. *   **XGBoost:** An optimized and highly efficient implementation of gradient boosting, known for its performance and speed. It includes regularization which helps prevent overfitting.

Given the high performance observed and XGBoost's reputation for efficiency and handling complex datasets, it would be a strong candidate for real-world weather prediction.
